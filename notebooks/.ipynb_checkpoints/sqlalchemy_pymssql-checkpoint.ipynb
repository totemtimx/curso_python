## 2) Cadenas de conexión (URIs) y mejores prácticas

Formato general de **SQLAlchemy** con `pymssql`:
```
mssql+pymssql://USUARIO:PASS@HOST:PUERTO/BASE?charset=utf8
```

**Ejemplos:**

- Autenticación SQL Server (usuario/contraseña):
```
mssql+pymssql://sa:SuperSecreta123@192.168.1.60:1433/ventas?charset=utf8
```

- Usar variables de entorno (recomendado):
```python
import os
from sqlalchemy import create_engine

DB_USER = os.getenv("DB_USER", "sa")
DB_PASS = os.getenv("DB_PASS", "SuperSecreta123")
DB_HOST = os.getenv("DB_HOST", "localhost")
DB_PORT = os.getenv("DB_PORT", "1433")
DB_NAME = os.getenv("DB_NAME", "ventas")

engine = create_engine(
    f"mssql+pymssql://{DB_USER}:{DB_PASS}@{DB_HOST}:{DB_PORT}/{DB_NAME}?charset=utf8",
    pool_pre_ping=True,       # valida conexiones del pool
    pool_size=5, max_overflow=10
)
```
> **Tip:** No guardes credenciales en el código. Usa variables de entorno o un `dotenv`/secret manager.

## 3) SQLAlchemy Core: conectar, ejecutar SQL y crear tablas

Con **Core** trabajas con conexiones explícitas, SQL textual y objetos `Table/Column`.


````

from sqlalchemy import create_engine, text, MetaData, Table, Column
from sqlalchemy import Integer, String, Numeric, DateTime, func

# ⚠️ Ajusta las variables de entorno antes de ejecutar este notebook contra tu servidor real.
import os
DB_USER = os.getenv("DB_USER", "sa")
DB_PASS = os.getenv("DB_PASS", "SuperSecreta123")
DB_HOST = os.getenv("DB_HOST", "localhost")
DB_PORT = os.getenv("DB_PORT", "1433")
DB_NAME = os.getenv("DB_NAME", "ventas_demo")

engine = create_engine(
    f"mssql+pymssql://{DB_USER}:{DB_PASS}@{DB_HOST}:{DB_PORT}/{DB_NAME}?charset=utf8",
    pool_pre_ping=True,
    pool_size=5, max_overflow=10
)

metadata = MetaData(schema=None)  # usa 'dbo' por defecto en MSSQL

productos = Table(
    "productos",
    metadata,
    Column("id", Integer, primary_key=True, autoincrement=True),
    Column("sku", String(50), unique=True, nullable=False),
    Column("nombre", String(200), nullable=False),
    Column("precio", Numeric(12, 2), nullable=False),
    Column("creado_en", DateTime, server_default=func.getdate()),  # GETDATE() de MSSQL
)

# Crea la(s) tabla(s) en la BD (si no existen)
# metadata.create_all(engine)
print("Definición de tabla preparada. Descomenta metadata.create_all(engine) para crear en tu BD real.")
````

## 4) SQLAlchemy ORM: modelos, sesiones y CRUD

El ORM te permite mapear clases Python a tablas y trabajar con `Session`.


````

from sqlalchemy.orm import DeclarativeBase, Mapped, mapped_column, Session

class Base(DeclarativeBase):
    pass

class Producto(Base):
    __tablename__ = "productos"
    # __table_args__ = {"schema": "dbo"}  # si requieres un esquema explícito

    id: Mapped[int] = mapped_column(primary_key=True, autoincrement=True)
    sku: Mapped[str] = mapped_column(String(50), unique=True, nullable=False)
    nombre: Mapped[str] = mapped_column(String(200), nullable=False)
    precio: Mapped[float] = mapped_column(Numeric(12,2), nullable=False)
    creado_en: Mapped[DateTime] = mapped_column(DateTime, server_default=func.getdate())

# Base.metadata.create_all(engine)  # crea tablas a partir de modelos
print("Modelo ORM definido. Descomenta Base.metadata.create_all(engine) para crear la tabla.")
````

## 8) Pooling (conexiones) y configuración

El **pool** reutiliza conexiones para mejorar rendimiento. Parámetros útiles en `create_engine`:
- `pool_size=5`: conexiones persistentes en el pool.
- `max_overflow=10`: conexiones extra cuando se satura el pool.
- `pool_timeout=30`: segundos a esperar antes de lanzar error por pool agotado.
- `pool_recycle=1800`: recicla conexiones inactivas (mitiga timeouts de firewalls).
- `pool_pre_ping=True`: verifica que la conexión esté viva antes de usarla.

```python
engine = create_engine(
    url,
    pool_size=5,
    max_overflow=10,
    pool_timeout=30,
    pool_recycle=1800,
    pool_pre_ping=True,
)
```

## 10) Esquemas, tipos y consideraciones de MSSQL

- Esquema por defecto: **dbo**. Si usas otros esquemas, especifica `schema="ventas"` al definir `Table` o en `__table_args__` del ORM.
- Fechas: usa `DateTime` y `func.getdate()` o `func.sysdatetime()` para defaults del lado servidor.
- Tipos numéricos: `Numeric(precision, scale)` evita problemas de redondeo en precios.
- Identidades: en ORM usa `autoincrement=True` y `primary_key=True`.
- Palabras reservadas: evita nombres de columna reservados o escápalos con `quoted_name` si es necesario.



<a id='checklist'></a>
## 11) Checklist de buenas prácticas y problemas comunes

**Credenciales y seguridad**
- ❒ Usa variables de entorno o un gestor de secretos (no subas contraseñas a Git).
- ❒ Limita permisos del usuario en la BD (principio de mínimo privilegio).
- ❒ Cierra sesiones/conexiones (usa `with` y `Session` como contexto).

**Conexión y red**
- ❒ Verifica reachability (`telnet HOST 1433` / `nc -vz HOST 1433` desde el servidor).
- ❒ Si hay proxy/firewall, pide una excepción para tu IP/puerto o usa túneles temporales.
- ❒ Habilita `pool_pre_ping=True` para evitar conexiones muertas.

**Esquema y datos**
- ❒ Define índices y restricciones (`UNIQUE`, `CHECK`, `FKs`) en el lado de la BD.
- ❒ Valida datos de entrada en el código (tipos, rangos, nulos).

**Transacciones**
- ❒ Agrupa operaciones relacionadas en una misma transacción.
- ❒ Captura `IntegrityError` y registra contexto útil (SKU, usuario, etc.).

**Rendimiento**
- ❒ Evita `SELECT *` en producción; selecciona columnas necesarias.
- ❒ Usa paginación (`OFFSET/FETCH`) en listados grandes.
- ❒ Evita `to_sql` masivo sin plan; prefiere `executemany`/bulk si es crítico.

**Migraciones**
- ❒ Considera **Alembic** para versionar el esquema (migraciones reproducibles).
