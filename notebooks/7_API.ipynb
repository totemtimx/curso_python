# Fast API

FastAPI es un framework web moderno y de alto rendimiento para construir APIs con Python 3.7+, basado en type hints estándar de Python.

### Características principales:

* Rápido: Rendimiento comparable a NodeJS y Go
* Fácil: Diseñado para ser intuitivo y reducir tiempo de desarrollo
* Automático: Genera documentación interactiva automáticamente
* Type hints: Usa anotaciones de tipo de Python para validación

```python
#Aplicación Básica

from fastapi import FastAPI
import uvicorn
import threading

app = FastAPI()

@app.get("/")
def read_root():
    return {"mensaje": "Hola Mundo"}

@app.get("/items/{item_id}")
def read_item(item_id: int, q: str = None):
    return {"item_id": item_id, "q": q}
```

```python
# Iniciar servidor en background

def run_server():
    uvicorn.run(app, host="127.0.0.1", port=8000)

thread = threading.Thread(target=run_server, daemon=True)
thread.start()
```

### Métodos mas comunes

@app.get("/items/")      # Obtener datos

@app.post("/items/")     # Crear

@app.put("/items/{id}")  # Actualizar completo

@app.patch("/items/{id}") # Actualizar parcial

@app.delete("/items/{id}") # Eliminar

### Query Parameters vs Path Parameters



```python
# Path parameter (obligatorio)
@app.get("/users/{user_id}")
def get_user(user_id: int):
    return {"user_id": user_id}

# Query parameter (opcional)
@app.get("/search/")
def search(q: str = None, limit: int = 10):
    return {"q": q, "limit": limit}
```

### Ejemplo Práctico: CRUD Básico

```python
from fastapi import FastAPI, HTTPException
from pydantic import BaseModel

#app = FastAPI()

# Modelo
class Producto(BaseModel):
    nombre: str
    precio: float
    stock: int

# Base de datos simulada
productos_db = {}
contador = 0

@app.post("/productos/")
def crear_producto(producto: Producto):
    global contador
    contador += 1
    productos_db[contador] = producto
    return {"id": contador, **producto.dict()}

@app.get("/productos/{producto_id}")
def obtener_producto(producto_id: int):
    if producto_id not in productos_db:
        raise HTTPException(status_code=404, detail="Producto no encontrado")
    return productos_db[producto_id]

@app.get("/productos/")
def listar_productos():
    return productos_db

@app.delete("/productos/{producto_id}")
def eliminar_producto(producto_id: int):
    if producto_id not in productos_db:
        raise HTTPException(status_code=404, detail="Producto no encontrado")
    del productos_db[producto_id]
    return {"mensaje": "Producto eliminado"}
```
